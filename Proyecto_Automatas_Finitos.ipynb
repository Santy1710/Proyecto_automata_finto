```python
def es_cedula_valida(cadena):
    estado = "q0"

    for c in cadena:
        if estado == "q0":
            if c.isdigit():
                estado = "q1"
            else:
                return False

        elif estado == "q1":
            if c == '-':
                estado = "q2"
            else:
                return False

        elif estado == "q2":
            if c.isdigit():
                estado = "q3"
            else:
                return False

        elif estado in ["q3", "q4", "q5"]:
            if c.isdigit():
                estado = "q" + str(int(estado[1]) + 1)
            else:
                return False

        elif estado == "q6":
            if c == '-':
                estado = "q7"
            else:
                return False

        elif estado == "q7":
            if c.isdigit():
                estado = "q8"
            else:
                return False

        elif estado in ["q8", "q9", "q10"]:
            if c.isdigit():
                estado = "q" + str(int(estado[1:]) + 1)
            else:
                return False

        elif estado == "q11":
            return False  # ya no debería haber más caracteres

    return estado == "q11"


# Pruebas
cedulas = [
    "1-1234-5678",
    "2-0000-0000",
    "12-1234-5678",  # inválida
    "1-123-5678",    # inválida
]

for c in cedulas:
    print(c, "=>", es_cedula_valida(c))
```
